```python
import matplotlib.pyplot as plt
import json
from glob import glob
import os
```

```python
test_json = glob('../CRT_utils/CRT_weights_and_config/CRT_COCO_random/CRT bbox PE new json/*.json')
```

```python
len(test_json)
```

```python
output_dir = '../CRT_utils/CRT_weights_and_config/CRT_COCO_random/CRT bbox PE new json'
```

```python
accuracies = []



for i in range (1, len(test_json)+1):
    
    json_path = os.path.join(output_dir, f'test_accuracies_epoch_{i}.json')
    
    
    with open(json_path, 'rb') as f:
        accuracy_dict = json.load(f)
        
        accuracies.append(accuracy_dict['total_accuracy'])
```

```python
plt.plot(accuracies)
```

```python
for i in range (1, len(test_json)+1):
    
    if accuracies[i-1] > 0.849:
        print(i)
```
