```python
import os
import json
import matplotlib.pyplot as plt

from glob import glob
```

```python
CRT_bbox_dir = '../CRT_utils/CRT_weights_and_config/CRT_COCO_random/CRT bbox PE new json/'
CRT_no_bbox_attn_dir = '../CRT_utils/CRT_weights_and_config/CRT_COCO_random/CRT no bbox PE new json/'
```

```python
bbox_accuracy_ls    = []
no_bbox_accuracy_ls = []


for i in range(1, 25):
    
    with open(os.path.join(CRT_bbox_dir, f'test_accuracies_epoch_{i}.json'), 'rb') as file:
        CRT_bbox_info = json.load(file)
        bbox_accuracy_ls.append(CRT_bbox_info['total_accuracy'])

    with open(os.path.join(CRT_no_bbox_attn_dir, f'test_accuracies_epoch_{i}.json'), 'rb') as file:
        CRT_no_bbox_info = json.load(file)
        no_bbox_accuracy_ls.append(CRT_no_bbox_info['total_accuracy'])
    
```

```python
plt.xticks(range(1, 25))
plt.ylim([0, 1.0])

plt.plot(bbox_accuracy_ls, label='CRT with bbox PE')
plt.plot(no_bbox_accuracy_ls, label='CRT no bbox PE (attention PE)')
plt.xlabel('Epochs', fontsize=14)
plt.ylabel('Accuracy', fontsize=14)
plt.legend()
```
