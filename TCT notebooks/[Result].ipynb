```python
import numpy as np
import pickle

import sys

sys.modules["numpy.core"] = np._core
sys.modules["numpy.core.numeric"] = np._core.numeric

import matplotlib.pyplot as plt
```

```python
SCEGRAM_result_CRT_bbox    = '../results/NGTU[SCEGRAM_CRT_bbox_PE]performance.pkl'
SCEGRAM_result_CRT_no_bbox = '../results/NGTU[SCEGRAM_CRT_no_bbox_PE]performance.pkl'
SCEGRAM_result_RANDOM      = '../results/NGTU[SCEGRAM_CRT_no_bbox_PE_RANDOM]performance.pkl'

COCO18_result_CRT_bbox    = '../results/NGTU[coco18_CRT_bbox_PE]performance.pkl'
COCO18_result_CRT_no_bbox = '../results/NGTU[coco18_CRT_no_bbox_PE]performance.pkl'
COCO18_result_RANDOM      = '../results/NGTU[coco18_CRT_bbox_PE_RANDOM]performance.pkl'

natural_result_CRT_bbox    = '../results/NGTU1[natural_CRT_bbox_PE]performance.pkl'
natural_result_CRT_no_bbox = '../results/NGTU1[natural_CRT_no_bbox_PE]performance.pkl'
natural_result_RANDOM      = '../results/NGTU1[natural_CRT_bbox_PE_RANDOM]performance.pkl'
```

```python
results = {}

with open(SCEGRAM_result_CRT_bbox, 'rb') as file:
    results['SCEGRAM_bbox'] = pickle.load(file)

with open(SCEGRAM_result_CRT_no_bbox, 'rb') as file:
    results['SCEGRAM_no_bbox'] = pickle.load(file)
    
with open(SCEGRAM_result_RANDOM, 'rb') as file:
    results['SCEGRAM_RANDOM'] = pickle.load(file)

with open(COCO18_result_CRT_bbox, 'rb') as file:
    results['COCO18_bbox'] = pickle.load(file)

with open(COCO18_result_CRT_no_bbox, 'rb') as file:
    results['COCO18_no_bbox'] = pickle.load(file)

with open(COCO18_result_RANDOM, 'rb') as file:
    results['COCO18_RANDOM'] = pickle.load(file)

with open(natural_result_CRT_bbox, 'rb') as file:
    results['natural_bbox'] = pickle.load(file)

with open(natural_result_CRT_no_bbox, 'rb') as file:
    results['natural_no_bbox'] = pickle.load(file)

with open(natural_result_RANDOM, 'rb') as file:
    results['natural_RANDOM'] = pickle.load(file)
```

```python
plt.xticks(range(1, 21))
plt.ylim([-0.05, 1.05])

plt.plot(results['COCO18_bbox']['TCT'],    label='COCO18 CRT bbox')
plt.plot(results['COCO18_no_bbox']['TCT'], label='COCO18 CRT no bbox')
plt.plot(results['COCO18_RANDOM']['TCT'],  label='COCO18 RANDOM')

plt.xlabel('Fixation number', fontsize=14)
plt.ylabel('Cumulative performance', fontsize=14)

plt.legend()
```
