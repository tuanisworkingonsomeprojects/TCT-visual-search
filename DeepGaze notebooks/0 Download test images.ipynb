```python
import os
import json
import requests
from tqdm import tqdm

image_dir = '../datasets/COCO18_dset_for_CRT_training/coco18_test_deepgaze/'

test_json = '../datasets/COCO18_dset_for_CRT_training/coco18_test_deepgaze.json'

val_url = 'http://images.cocodataset.org/val2017'
train_url = 'http://images.cocodataset.org/train2017'
```

```python
with open(test_json, 'rb') as file:
    json_info = json.load(file)

images = json_info['images']

for image_info in tqdm(images):

    if os.path.exists(os.path.join(image_dir, image_info['file_name'])):
        continue


    url = val_url 

    image_url = os.path.join(url, image_info['file_name'])

    res  = requests.get(image_url, stream=True)

    if res.status_code == 404:
        url = train_url

        image_url = os.path.join(url, image_info['file_name'])

        res  = requests.get(image_url, stream=True)

    res.raise_for_status()

    with open(os.path.join(image_dir, image_info['file_name']), 'wb') as file:
        file.write(res.content)
```
