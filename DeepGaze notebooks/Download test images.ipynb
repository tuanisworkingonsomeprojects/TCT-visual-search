```python
import os
import json
import requests
from tqdm import tqdm

image_dir = '../datasets/COCO18_dset_for_CRT_training/coco18_test_deepgaze/'

test_json = '../datasets/COCO18_dset_for_CRT_training/coco18_test_deepgaze.json'

val_url = 'http://images.cocodataset.org/val2017'
train_url = 'http://images.cocodataset.org/train2017'
```

```python
with open(test_json, 'rb') as file:
    json_info = json.load(file)

images = json_info['images']

for image_info in tqdm(images):

    if os.path.exists(os.path.join(image_dir, image_info['file_name'])):
        continue


    url = val_url 

    image_url = os.path.join(url, image_info['file_name'])

    res  = requests.get(image_url, stream=True)

    if res.status_code == 404:
        url = train_url

        image_url = os.path.join(url, image_info['file_name'])

        res  = requests.get(image_url, stream=True)

    res.raise_for_status()

    with open(os.path.join(image_dir, image_info['file_name']), 'wb') as file:
        file.write(res.content)
```

```python
with open('../datasets/COCO18_dset_for_CRT_training/train.json', 'rb') as file:
    annotation_train_json_info = json.load(file)

    image_ls = []
    image_id_ls = []

    for image_info in annotation_train_json_info['images']:
        # Format from normal integer.jpg to leading 0000...jpg file format
        image_info['file_name'] = f"{int(image_info['file_name'].split('.')[0]):012}.jpg"

        if image_info['id'] not in image_id_ls:
            image_id_ls.append(image_info['id'])
            image_ls.append(image_info)
            

    annotation_train_json_info['images'] = image_ls

with open('../datasets/COCO18_dset_for_CRT_training/val.json', 'rb') as file:
    annotation_val_json_info = json.load(file)

    image_ls = []
    image_id_ls = []

    for image_info in annotation_val_json_info['images']:
        # Format from normal integer.jpg to leading 0000...jpg file format
        image_info['file_name'] = f"{int(image_info['file_name'].split('.')[0]):012}.jpg"

        if image_info['id'] not in image_id_ls:
            image_id_ls.append(image_info['id'])
            image_ls.append(image_info)
            

    annotation_val_json_info['images'] = image_ls

annotation = {}
annotation['info']        = annotation_train_json_info['info']
annotation['licenses']    = annotation_train_json_info['licenses']
annotation['images']      = annotation_train_json_info['images']      + annotation_val_json_info['images']
annotation['annotations'] = annotation_train_json_info['annotations'] + annotation_val_json_info['annotations']

```

```python
file_name_height_width = {
    x['file_name']: {
        'height': x['height'],
        'width' : x['width']
    } for x in annotation['images']}
```

To find the correct bbox for the annotation we need to devide each dimension for height and width of the original annotations and then we try to multipy with the image width and height

```python
for image_info in json_info['images']:
    image_info['height'] = file_name_height_width[image_info['file_name']]['height']
    image_info['width']  = file_name_height_width[image_info['file_name']]['width']
```

```python
deepgazefile_name_id = {x['file_name']: x['id'] for x in json_info['images']}
```

```python
file_name_id_annotation = {x['file_name']: x['id'] for x in annotation['images']}
id_bbox_annotation = {x['image_id']: x['bbox'] for x in annotation['annotations']}
file_name_bbox = {x : id_bbox_annotation[file_name_id_annotation[x]] for x in file_name_id_annotation if file_name_id_annotation[x] in id_bbox_annotation}
```

```python
deepgazefile_id_bbox = {deepgazefile_name_id[x]: file_name_bbox[x] for x in deepgazefile_name_id}
```

```python
for annotation_info in json_info['annotations']:
    annotation_info['bbox'] = deepgazefile_id_bbox[annotation_info['image_id']]
```

```python
with open(test_json, 'w') as file:
    json.dump(json_info, file)
```
